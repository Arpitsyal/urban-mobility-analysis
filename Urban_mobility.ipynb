NUMERICAL DATA QUALITY AUDIT

```python
numeric_columns = [
    "Boardings",
    "Alightings",
    "Load",
    "Capacity",
    "Seats",
    "V/C"
]

for col in numeric_columns:
    print(f"\n--- {col} ---")
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Mean:", df[col].mean())
    print("Zero values:", (df[col] == 0).sum())
    print("Negative values:", (df[col] < 0).sum())
```

```python
df["Calculated_VC"] = df["Load"] / df["Capacity"]

df["VC_Difference"] = abs(df["V/C"] - df["Calculated_VC"])

print("Maximum difference between reported V/C and calculated V/C:",
      df["VC_Difference"].max())

print("Rows with difference greater than 0.001:",
      (df["VC_Difference"] > 0.001).sum())
```

stastical outlier audit

```python
numeric_columns = [
    "Boardings",
    "Alightings",
    "Load",
    "Capacity",
    "Seats",
    "V/C"
]

print("Skewness of numerical variables:\n")

for col in numeric_columns:
    print(f"{col}: {df[col].skew():.3f}")
```

outlier check

```python
for col in ["Boardings", "Alightings", "Load", "Capacity", "Seats", "V/C"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}:")
    print(f"  Lower bound: {lower:.2f}")
    print(f"  Upper bound: {upper:.2f}")
    print(f"  Potential outliers: {outliers}")
```

OG DATASET FRESH RELOAD

```python
import pandas as pd

file_path = "/content/2025_26 Weekday TOTAL DEMAND BY ROUTE BY QUARTER HOUR Routes 1-149.csv"

df = pd.read_csv(file_path)

print("Dataset reloaded.")
print("Shape:", df.shape)
print("Number of unique QHr:", df["QHr"].nunique())
print("Number of unique TIMEBAND:", df["TIMEBAND"].nunique())
print("Number of unique routes:", df["ROUTE"].nunique())
```

```python
df_raw = pd.read_csv(file_path)
df = df_raw.copy()
```

```python
# Keep an untouched copy of the original dataset
df_raw = df.copy()

# Working copy for cleaning and preparation
df = df_raw.copy()

print("Raw dataset shape:", df_raw.shape)
print("Working dataset shape:", df.shape)
```

```python
# Convert QHr to time
df["QHr"] = pd.to_datetime(
    df["QHr"],
    format="%H:%M:%S"
).dt.time

# Convert STOPSEQUENCE to integer
df["STOPSEQUENCE"] = df["STOPSEQUENCE"].astype(int)

print(df.dtypes)
```

```python
print("Rows:", len(df))
print("Unique QHr:", df["QHr"].nunique())
print("Unique TIMEBAND:", df["TIMEBAND"].nunique())
print("Unique Routes:", df["ROUTE"].nunique())
```

```python
# Check whether each STOPCODE is associated with only one STOPNAME

stopcode_name_check = (
    df.groupby("STOPCODE")["STOPNAME"]
      .nunique()
      .sort_values(ascending=False)
)

print("Maximum number of different STOPNAMEs for one STOPCODE:",
      stopcode_name_check.max())

print("\nSTOPCODES associated with more than one STOPNAME:")
print(stopcode_name_check[stopcode_name_check > 1].head(20))
```

```python
stopname_code_check = (
    df.groupby("STOPNAME")["STOPCODE"]
      .nunique()
      .sort_values(ascending=False)
)

print("Maximum number of STOPCODES for one STOPNAME:",
      stopname_code_check.max())

print("\nSTOPNAMEs associated with multiple STOPCODEs:")
print(stopname_code_check[stopname_code_check > 1].head(20))
```

```python
park_road = (
    df[df["STOPNAME"] == "PARK ROAD"]
    [["STOPCODE", "STOPNAME"]]
    .drop_duplicates()
    .sort_values("STOPCODE")
)

print("Different STOPCODEs for PARK ROAD:")
display(park_road)
```

CREATE HOUR FIELD

```python
# Create an hourly field from the quarter-hour field
df["Hour"] = [t.hour for t in df["QHr"]]

print(df[["QHr", "Hour"]].head(10))
print("\nUnique hours:", sorted(df["Hour"].unique()))
```

```python
# Calculate the net change in passenger count at each stop observation
df["Net_Passenger_Change"] = df["Boardings"] - df["Alightings"]

print(
    df[
        ["Boardings", "Alightings", "Net_Passenger_Change", "Load"]
    ].head(10)
)
```

```python
print("Positive net passenger change:",
      (df["Net_Passenger_Change"] > 0).sum())

print("Negative net passenger change:",
      (df["Net_Passenger_Change"] < 0).sum())

print("Zero net passenger change:",
      (df["Net_Passenger_Change"] == 0).sum())

print("\nMinimum:",
      df["Net_Passenger_Change"].min())

print("Maximum:",
      df["Net_Passenger_Change"].max())
```

```python
df.groupby("ROUTE")["Boardings"].sum()
```

```python
key_columns = [
    "YEAR",
    "DAY_TYPE",
    "QHr",
    "ROUTE",
    "DIRECTION",
    "STOPCODE",
    "STOPSEQUENCE"
]

key_duplicates = df.duplicated(
    subset=key_columns
).sum()

print("Duplicate records based on operational key:", key_duplicates)
```

```python
print("Net Passenger Change summary:")
print(df["Net_Passenger_Change"].describe())

print("\nTop 10 positive net changes:")
display(
    df.nlargest(10, "Net_Passenger_Change")[
        ["QHr", "ROUTE", "DIRECTION", "STOPCODE",
         "STOPNAME", "Boardings", "Alightings",
         "Net_Passenger_Change", "Load", "Capacity", "V/C"]
    ]
)

print("\nTop 10 negative net changes:")
display(
    df.nsmallest(10, "Net_Passenger_Change")[
        ["QHr", "ROUTE", "DIRECTION", "STOPCODE",
         "STOPNAME", "Boardings", "Alightings",
         "Net_Passenger_Change", "Load", "Capacity", "V/C"]
    ]
)
```
